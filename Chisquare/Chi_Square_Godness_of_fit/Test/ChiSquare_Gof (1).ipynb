# Business Problem :

A university claims that its baseline testing system is designed so that exactly 65% of students should pass any given exam, while 35% should fail. An auditor wants to verify if the actual exam results follow this expected distribution. They pull a random sample of 100 student records from the database.

The raw data shows the following final counts:

• Pass: 55 students (40 Male + 15 Female)

• Fail: 45 students (20 Male + 25 Female)

# Business Statement
Using a significance level of α = 0.05, conduct a Chi-Square Goodness-of-Fit test to determine if the observed distribution of exam results fits the university's claimed distribution.

H0 : Actual exam result follow the expected distribution

H1 : Actual Exam result do not Follow the Expected Distribution

Test : Godness of Fit

```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_excel('/content/Chi_Square_Degree_of_Association_Practice.xlsx')
df.head()
```

```python
import seaborn as sns

sns.countplot(data=df,x='Gender',hue='Result')
```

```python
observed=[55,45]

expected=[0.65,0.35]

pattern=sum(observed)
pattern

expected=[pattern*0.65,pattern*0.35]
expected

```

```python
from statsmodels.stats.api import gof

stats,p_value=gof.chisquare(observed,expected)
```

```python
stats
```

```python
p_value
```

```python
if(p_value<0.05):
  print("Reject the Null hypothesis")
else:
  print("we Fail to Reject the null Hypothesis")
```
