Context: The Customer Success department operates under a strict executive Service Level Agreement (SLA) stating that the annual customer churn rate must not exceed 5.0%. If the true churn rate rises above this threshold, the company must immediately reallocate budget to emergency retention campaigns.

Business Question: Based on the performance metrics in the dataset, is the platform's actual subscription churn rate (Subscription_Churned) significantly higher than the mandated 5.0% target?

H0:Actual Subscription churn rate is not higher than the mandate 5.0% target

H1 : Actual Subscription Churn Rate is Higher thn the mandate 5.0% target

```python
import numpy as np 
import pandas as pd 
```

```python
df=pd.read_csv('business_proportion_test_data.csv')
```

```python
df.head()
```

```python
X=df['Subscription_Churned'].value_counts()

s=X/100
s

sum(df.Subscription_Churned)
```

```python
from statsmodels.stats.proportion import proportions_ztest
p0=df['Subscription_Churned'].sum()

nobs=df['Subscription_Churned'].count()

z_stats,p_value=proportions_ztest(p0,nobs,value=0.05,alternative='larger')

```

```python
p_value
```

```python
if(p_value<0.05):
    print("Reject null hypothesis ")
else:
    print("fail to reject hypothesis")
    
```

CONCLUSION :

Actual Subscription Churn Rate is Higher than the mandate 5.0% target
