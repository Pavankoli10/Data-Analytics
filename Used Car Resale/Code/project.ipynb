```python
# Independent variables
X = cdf.drop(
    columns=['selling_price', 'km_per_year', 'selling_price_log'],
    errors='ignore'
)

# Dependent variable
Y= cdf['selling_price']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression

categorical_features = [
    'brand', 'model', 'transmission', 'fuel', 'owner'
]

numerical_features = [
    'age', 'engine', 'km'
]

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', categorical_transformer, categorical_features),
        ('num', numerical_transformer, numerical_features)
    ]
)
```

```python
lr_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,Y,random_state=42,test_size=0.20)
lr_model.fit(x_train,y_train)
```

```python
y_pred=lr_model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

print("MAE:", mean_absolute_error(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
train_pred = lr_model.predict(x_train)

print("Training R2:", r2_score(y_train, train_pred))
print("Testing R2:", r2_score(y_test, y_pred))
```

***DECISION TREE***

```python
from sklearn.tree import DecisionTreeRegressor
```

```python

def dtmodel(depth=1):
    return Pipeline([
        ('preprocessor',preprocessor),
        ('dect_reg',DecisionTreeRegressor(max_depth=depth))
    ])

estimator=dtmodel()

from sklearn.model_selection import GridSearchCV


param_grid={
    'dect_reg__max_depth':[range(1,10),None]
}

grid=GridSearchCV(
    estimator=estimator,
    param_grid=param_grid,
    cv=3,
    scoring='accuracy'
)
grid.fit(x_train,y_train)
bestmodel=grid.best_estimator_
```

```python
dt_pred=bestmodel.predict(x_test)
```

```python
print("MAE:", mean_absolute_error(y_test, dt_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, dt_pred)))
print("R2 Score:", r2_score(y_test, dt_pred))
dt_train_pred = bestmodel.predict(x_train)
print("Training R2:", r2_score(y_train, dt_train_pred))
```

```python


from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# 1. Generate training predictions
lr_train_pred = lr_model.predict(x_train)
dt_train_pred = bestmodel.predict(x_train)
rf_train_pred = rf_model.predict(x_train)

# 2. Generate testing predictions
lr_test_pred = lr_model.predict(x_test)
dt_test_pred = bestmodel.predict(x_test)
rf_test_pred = rf_model.predict(x_test)

def evaluate_model(name, y_train, train_pred, y_test, test_pred):

    return {
        'Model': name,

        'MAE': mean_absolute_error(y_test, test_pred),

        'MSE': mean_squared_error(y_test, test_pred),

        'RMSE': np.sqrt(
            mean_squared_error(y_test, test_pred)
        ),

        'Training R2': r2_score(y_train, train_pred),

        'Testing R2': r2_score(y_test, test_pred)
    }

results = []

results.append(
    evaluate_model(
        'Linear Regression',
        y_train, lr_train_pred,
        y_test, lr_test_pred
    )
)

results.append(
    evaluate_model(
        'Decision Tree',
        y_train, dt_train_pred,
        y_test, dt_test_pred
    )
)

results.append(
    evaluate_model(
        'Random Forest',
        y_train, rf_train_pred,
        y_test, rf_test_pred
    )
)
comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by='Testing R2',
    ascending=False
).reset_index(drop=True)

display(comparison_df.round(4))
```

```python
best_model_name = comparison_df.loc[
    comparison_df['Testing R2'].idxmax(),
    'Model'
]

print("Best Model:", best_model_name)
```
