```python
for col in df.select_dtypes(include='number').columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\nColumn: {col}")
    print("Number of outliers:", len(outliers))
    print(outliers[col].values)
```

```python
cols = ['Retail_Price', 'Wholesale_Price', 'Inflation_Rate']

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower=lower, upper=upper)

print("Outliers capped successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a copy to preserve the original dataset
df = df.copy()

# Convert Date to datetime
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Convert price columns to numeric
price_cols = ['Retail_Price', 'Wholesale_Price']

for col in price_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Remove rows missing essential analysis values
df = df.dropna(
    subset=['Date', 'Commodity', 'State',
            'Retail_Price', 'Wholesale_Price']
)

# Create year-month and calendar-month columns
df['Year_Month'] = df['Date'].dt.to_period('M').astype(str)
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.strftime('%B')

# Set the plotting style
sns.set_theme(style='whitegrid')

print("Data prepared successfully!")
print(df.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.dates as mdates

# 1. Prepare the data
cdf['Date'] = pd.to_datetime(cdf['Date'], errors='coerce')
cdf['Retail_Price'] = pd.to_numeric(
    cdf['Retail_Price'], errors='coerce'
)

selected_commodities = ['Rice', 'Wheat', 'Tomato', 'Potato']

plot_df = cdf[
    cdf['Commodity'].isin(selected_commodities)
].copy()

plot_df = plot_df.dropna(
    subset=['Date', 'Retail_Price', 'Commodity']
)

# 2. Calculate monthly average price
plot_df['Month'] = (
    plot_df['Date'].dt.to_period('M').dt.to_timestamp()
)

monthly_prices = (
    plot_df.groupby(['Month', 'Commodity'], as_index=False)
    ['Retail_Price'].mean()
)

# 3. Set chart style
sns.set_theme(style='whitegrid', context='notebook')

# 4. Create a 2 × 2 grid
fig, axes = plt.subplots(2, 2, figsize=(18, 10))

axes = axes.flatten()

for i, commodity in enumerate(selected_commodities):

    ax = axes[i]

    commodity_df = monthly_prices[
        monthly_prices['Commodity'] == commodity
    ]

    if commodity_df.empty:
        ax.set_title(f'No data available for {commodity}')
        continue

    # Plot the line graph
    sns.lineplot(
        data=commodity_df,
        x='Month',
        y='Retail_Price',
        marker='o',
        markersize=3,
        linewidth=2,
        ax=ax
    )

    # Title and labels
    ax.set_title(
        f'Monthly Average Retail Price - {commodity}',
        fontsize=13,
        fontweight='bold',
        pad=10
    )

    ax.set_xlabel('Month', fontsize=11)
    ax.set_ylabel('Average Retail Price', fontsize=11)

    # Format dates
    ax.xaxis.set_major_locator(
        mdates.MonthLocator(interval=4)
    )
    ax.xaxis.set_major_formatter(
        mdates.DateFormatter('%b %Y')
    )

    ax.tick_params(axis='x', rotation=45)
    ax.grid(True, linestyle='--', alpha=0.4)

# 5. Add an overall title
fig.suptitle(
    'Monthly Retail Price Trends of Selected Commodities',
    fontsize=17,
    fontweight='bold',
    y=1.02
)

# 6. Adjust spacing
plt.tight_layout()
plt.show()
```

### Key Insights

1. **Rice:** Rice had the highest retail price among the selected commodities and showed an overall increasing trend from 2019 to 2026.

2. **Wheat:** Wheat prices increased gradually over the study period, with relatively fewer sharp fluctuations than tomato prices.

3. **Tomato:** Tomato prices showed frequent and sharp fluctuations, possibly due to seasonal availability, weather conditions, and changes in supply and demand.

4. **Potato:** Potato prices experienced repeated rises and falls, suggesting possible seasonal variations in price.

5. **Overall Finding:** Rice and wheat showed a general upward price trend, while tomato and potato experienced more noticeable fluctuations. These patterns highlight the importance of monitoring commodity-wise price changes over time.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Select numerical columns
numeric_cols = [
    'Retail_Price',
    'Wholesale_Price',
    'Inflation_Rate'
]

# Convert columns to numeric
for col in numeric_cols:
    cdf[col] = pd.to_numeric(cdf[col], errors='coerce')

# Calculate correlation
correlation = cdf[numeric_cols].corr(method='pearson')

# Create heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    center=0,
    linewidths=1,
    square=True,
    cbar_kws={'label': 'Correlation Coefficient'}
)

plt.title('Correlation Heatmap of Food Price Indicators',
          fontsize=15, fontweight='bold', pad=15)

plt.xticks(rotation=30, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert Retail_Price to numeric
cdf['Retail_Price'] = pd.to_numeric(cdf['Retail_Price'], errors='coerce')

# Remove missing values
plot_df = cdf.dropna(subset=['State', 'Commodity', 'Retail_Price']).copy()

# Select commodities
selected_commodities = ['Rice', 'Wheat', 'Tomato', 'Potato']
plot_df = plot_df[plot_df['Commodity'].isin(selected_commodities)]

# Calculate average retail price by state and commodity
state_prices = (
    plot_df.groupby(['State', 'Commodity'], as_index=False)
    ['Retail_Price'].mean()
)

# Plot grouped bar chart
plt.figure(figsize=(16, 8))

sns.barplot(
    data=state_prices,
    x='State',
    y='Retail_Price',
    hue='Commodity',
    order=sorted(state_prices['State'].unique()),
    hue_order=selected_commodities,
    errorbar=None
)

plt.title('State-wise Comparison of Average Retail Prices by Commodity',
          fontsize=16, fontweight='bold', pad=15)
plt.xlabel('State', fontsize=12)
plt.ylabel('Average Retail Price', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Commodity')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```
